# Dictionaries

A dictionary in Python is a built-in data structure used to store data in key-value pairs, where each unique key maps to a specific value. It is a mutable (changeable) collection that preserves its insertion order and does not allow duplicate keys.

user_dictionary = {
    'username': 'JohnCena',
    'name': 'John',
    'age': 32
}
# addes new entry to the dictionary
user_dictionary["married"] = True
# prints the full dictionary
print (user_dictionary)



# removing property 
user_dictionary.pop("age")
for x, y in user_dictionary.items():
    print(x, y)

# when items are assinged they are passed by reference
user_dictionary2 = user_dictionary
# this is passed by value
user_dictionary3= user_dictionary.copy()

# Dictionary Iteration

You can interate through the items in the dictionary using for loops.

```python
# prints the first item from the list
for x in user_dictionary:
    print(x)
```

```python
# prints the values
for val in user_dictionary.values():
    print(val)

```

```python
# Prints the keys
for key in user_dictionary.keys():
    print(key)
```

```python
# Print the keys and values
for (key, value) in user_dictionary.items():
    print (key, value)
```
