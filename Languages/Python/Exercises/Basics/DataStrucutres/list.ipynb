# List

Lists are a collection of data. Similar to arrays.

```python
# Prints the list
my_list = [80,96,72,100,8]
print(my_list)

people_list = ["John", "Jane","Ozzy","Batman"]
print(people_list)
```

# Iterating though the list

We can use for loops or while loops to iterate through the list. Below are two examples using for loop and while loop.

```python
#example using while loop
index = 0
while (index < len(my_list)):
    print(my_list[index])
    index+=1
```

```python
# Using for loop to list out the item in people_list
for people in people_list:
    print(people)
```

In Python if you put -1 as the index it will always grab the last element of the item

```python
print(people_list[-1])
print(my_list[-1])
```

# Slicing

The following command will display the list of items from my_list where the index start at 2 and ends at 4 but does not include the item from index 4

```python
print(my_list[2:4])
```

# List Mutation

Items can be added or removed from the lists using the following predefined methods available on the list.
- append : adds at the end
- insert : adds at the given index 
- remove : removes the give value
- pop : removes the last item

```python
# Sample for append and insert
mutation_list = [80,96,72,100,8]
# inserts 27 as the 3rd element on the list at index 2
mutation_list.insert(2, 27)
# inserts 99 as the last element on the list
mutation_list.append(99)
# displays the full list
print(mutation_list)
```

```python
# Sample for remove and pop
mutation_list = [80,96,72,100,8]

print("Original list")
print(mutation_list)

mutation_list.remove(8)
print("Displays the list with 8 removed")
print(mutation_list)

mutation_list.pop()
print("Last element removed removed")
print(mutation_list)
```

# Sorting

Lists can be sorted using .sort on list variable. In order for the sort to work, items in the lists needs to be of same data type.

```python
unsorted_list = [80,96,72,100,8]
unsorted_list.sort()
print(unsorted_list)

people_list.sort()
print(people_list)
```

## Extend

extend method can be used to connect the two lists. Below is the sample where my_list will be extended with people_list

```python
list_1 = [80,96,72,100,8]
list_2=["Apple", "Banana", "Orange"]
list_1.extend(list_2)
print(list_1)
```
