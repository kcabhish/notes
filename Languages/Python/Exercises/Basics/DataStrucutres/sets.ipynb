# Sets

Sets are similar to listsbut are unordered and cannont contain duplicate items. Use Curly Brackets.

```python
my_set = {1,2,3,4,5,1,2}
print("Sets automatically removes duplicate items")
print(my_set)
print("Length function will not include duplicate items so it will return the length with unique data")
print(len(my_set))
```

# Iterating Sets

Sets can be iterated through the for loop. 

NOTE: you cannont use index for this as the sets are unordered in memory

```python
for item in my_set:
    print(item)
```

# Set Mutation

The following functions can be used to mutate the sets.

- discard() : revmoes a value  into the set.
- add() : adds a value into the set.
- update() : adds multiple values into the set.


```python
# Adding single item to the list
example_list1 = {2,4,8,1,17,4}
print(example_list1)
example_list1.add(7)
print(example_list1)
example_list1.remove(17)
print(example_list1)
example_list1.update([22,23])
print(example_list1)
```
