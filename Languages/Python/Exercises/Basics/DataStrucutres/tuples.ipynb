# Tuples

A tuple is a built-in data type used to store an ordered, immutable (unchangeable) collection of items. Think of a tuple like a list that is locked in place after creation; you can access its contents, but you cannot add, remove, or modify elements


```python
# Creating a tuple
user_profile = ("Alice", 28, "Engineer", True)

# Accessing items via indexing
print(user_profile[0])  # Outputs: Alice
print(user_profile[1])  # Outputs: 28

# Slicing a tuple
print(user_profile[1:3])  # Outputs: (28, 'Engineer')
```
