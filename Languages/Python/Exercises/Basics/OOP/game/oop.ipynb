## What is Abstraction?

Abstraction means to hide the implementation details and only expose what is necessary detals to the user.

## Why use Abstraction?

- This allows users to not have to unerstand what the functionality is behind scenes.
- You can create simple and reusable code.
- Allows for a better use of the DRY principle (Don't repeat yourself)
- Enbles Pyhthon objcts to becoome more scalable

## What is Encapsulation

Enacapsulation means the bundling of data. This prevents users from accessing or modifying object properties after the class is instantiated. Basically the private and public concept. In Python private properties are declared by prefixing variables with `__` (Double underscore)

## Why use Encapsulation

- Helps keep related fields and methods together.
- Makes our code cleaner and easier to read.
- Provides more flexibility to our code.
- Provides more reusability with our code.

### Default/Empty Constructors
Below is an example of the default / empty construtors. This class can be instantiated using `enemy = Enemy()`

```python

class Enemy:
    type_of_enemy: str
    health_points: int=10
    attack_damage: int=1

    # This is a default construtor automatically created if no constructor is found. 
    # the below section has been manually added to represetn an example but in actuality this will be hidden for default.
    def __init__(self):
        pass

    def talk(self):
        print(f'I am a {self.type_of_enemy}. Be prepared to fight')

    def walk_forward(self):
        print(f'{self.type_of_enemy} moves close to you')
    
    def attack(self):
        print(f'{self.type_of_enemy} attacks for {self.attack_damage}')

#instantiating the Class
enemy = Enemy()
enemy.type_of_enemy='Vampires'
enemy.talk()
enemy.walk_forward()
enemy.attack()
```

### No Argument Construtors

Similar to default constructors, no arguement constructors can be used to overwrite the default constructor and instead of the pass, it can be modifed.

```python

class Enemy:
    type_of_enemy: str
    health_points: int=10
    attack_damage: int=1

    def __init__(self):
        print('New enemy created with no starting values')

    def talk(self):
        print(f'I am a {self.type_of_enemy}. Be prepared to fight')

    def walk_forward(self):
        print(f'{self.type_of_enemy} moves close to you')
    
    def attack(self):
        print(f'{self.type_of_enemy} attacks for {self.attack_damage}')

#instantiating the Class
enemy = Enemy()
enemy.type_of_enemy='Vampires'
enemy.talk()
enemy.walk_forward()
enemy.attack()
```

### Parameter Constructors

Users can pass in parameterized constructors which can be use to update private variables  based on the object instnatiated from the user.

```python

class Enemy:
    type_of_enemy: str
    health_points: int=10
    attack_damage: int=1

    def __init__(self):
        print('New enemy created with no starting values')
    
    def __init__(self, type_of_enemy, health_points=10, attack_damage=1):
        self.type_of_enemy: str = type_of_enemy
        self.health_points: int=health_points
        self.attack_damage: int=attack_damage
        print(f'New enemy {self.type_of_enemy} created with {self.attack_damage} attack values!')

    def talk(self):
        print(f'I am a {self.type_of_enemy}. Be prepared to fight')

    def walk_forward(self):
        print(f'{self.type_of_enemy} moves close to you')
    
    def attack(self):
        print(f'{self.type_of_enemy} attacks for {self.attack_damage}')

#instantiating the Class
warewolf = Enemy('warewolf', 14, 2)
ogre = Enemy('Ogre', 12, 1)
```

# Encapsulation Example

Below is the code snippet where type_of_enemy has been instantiated as a private property in the Enemy class. The property can now only be accessed within the class or through a public methods from an object. In this case its the getter method for the property.

```python
class Enemy:
    __type_of_enemy: str
    health_points: int=10
    attack_damage: int=1

    def __init__(self):
        print('New enemy created with no starting values')
    
    def __init__(self, type_of_enemy, health_points=10, attack_damage=1):
        self.__type_of_enemy: str = type_of_enemy
        self.health_points: int=health_points
        self.attack_damage: int=attack_damage
        print(f'New enemy {self.__type_of_enemy} created with {self.attack_damage} attack values!')

    def talk(self):
        print(f'I am a {self.__type_of_enemy}. Be prepared to fight')

    def walk_forward(self):
        print(f'{self.__type_of_enemy} moves close to you')
    
    def attack(self):
        print(f'{self.__type_of_enemy} attacks for {self.attack_damage}')

    def get_type_of_enemy(self):
        return self.__type_of_enemy
    
# Instantiating object
warewolf = Enemy('warewolf', 14, 2)
print(warewolf.get_type_of_enemy())
```
