## Abstraction

### What is Abstraction?

Abstraction means to hide the implementation details and only expose what is necessary detals to the user.

### Why use Abstraction?

- This allows users to not have to unerstand what the functionality is behind scenes.
- You can create simple and reusable code.
- Allows for a better use of the DRY principle (Don't repeat yourself)
- Enbles Pyhthon objcts to becoome more scalable

## Encapsulation

### What is Encapsulation?

Enacapsulation means the bundling of data. This prevents users from accessing or modifying object properties after the class is instantiated. Basically the private and public concept. In Python private properties are declared by prefixing variables with `__` (Double underscore)

### Why use Encapsulation?

- Helps keep related fields and methods together.
- Makes our code cleaner and easier to read.
- Provides more flexibility to our code.
- Provides more reusability with our code.

### Encapsulation Example

Below is the code snippet where type_of_enemy has been instantiated as a private property in the Enemy class. The property can now only be accessed within the class or through a public methods from an object. In this case its the getter method for the property.

```python
class Enemy:
    __type_of_enemy: str
    health_points: int=10
    attack_damage: int=1

    def __init__(self):
        print('New enemy created with no starting values')
    
    def __init__(self, type_of_enemy, health_points=10, attack_damage=1):
        self.__type_of_enemy: str = type_of_enemy
        self.health_points: int=health_points
        self.attack_damage: int=attack_damage
        print(f'New enemy {self.__type_of_enemy} created with {self.attack_damage} attack values!')

    def talk(self):
        print(f'I am a {self.__type_of_enemy}. Be prepared to fight')

    def walk_forward(self):
        print(f'{self.__type_of_enemy} moves close to you')
    
    def attack(self):
        print(f'{self.__type_of_enemy} attacks for {self.attack_damage}')

    def get_type_of_enemy(self):
        return self.__type_of_enemy
    
# Instantiating object
warewolf = Enemy('warewolf', 14, 2)
print(warewolf.get_type_of_enemy())
```

## Inheritance

## What is Inheritance?

Process of acquiring properties from one class to another classes. Inheritance allows chlid classes to access the public properties from the parent. Child classes can override the parent properties.

## Why use Inheritance?

- Eliminates duplication: Write common logic once in a parent class.Automatic updates: Changes to parent code automatically update child classes.Saves time: Build on existing, pre-tested code instead of starting over.


```python
# Inheritance example
# Parent
class Animal:
    weight: int
    color: str
    age: int
    animal_type: str

    def eat(self):
        print(f'{self.animal_type} eating!')
    
    def sleep(self):
        print(f'{self.animal_type} Sleeping!')

    def talk(self):
        print("value is not set for this animal")

# Child
class Dog(Animal):
    ## includes all animal attributes
    can_shed: bool
    domestic_name: str
    animal_type = "dog"

    def talk(self):
        print('Bark!')

class Bird(Animal):
    birdType: str

    def talk(self):
        print('chirp!')
    
    def fly(self):
        print('Bird begins to soar!')

# Creating instance of a dog

dog_obj = Dog()
dog_obj.eat();
dog_obj.sleep();

bird_obj = Bird()
bird_obj.talk();
bird_obj.fly();
```

### self vs super

In OOP Inheritance, when referencing properties inside the same class, use self, when referencing properties of the parent or the base class use super.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age: int = age

class Student(Person):
    def __init__(self, name, age, degree):
        super().__init__(name, age)
        self.degree = degree
    
    def display_student(self):
        print(f'{self.name}, {self.age} from parent, and {self.degree} from child')

student = Student('John', 12, 'CS')
student.display_student()
```

## Polymorphism

Polymorphism in Python is a core Object-Oriented Programming (OOP) concept that translates to "many forms". It refers to the ability of different objects to respond to the exact same method name, function, or operator in their own unique ways. Essentially, it lets you establish a single uniform interface for a variety of underlying data types or classes.

## Composition

- Its a way to create objects made up of other objects.
- In composition, a class contains one or more objects of another class as istance variables.
- Provide layered functionality to the object.
- Known as a HAS-A relationship.

In the example below, it is using a composition bcause the vehicle HAS-A engine.
This is different from an interface IS-A relationship.
A vehicle must have an engine but an engine does not need to have a vehilce.

```python
class Engine:
    def __init__(self, engineType):
        self.engineType = engineType
    
    def startEngine(self):
        print('Engine is running!')

    def stopEngine(self):
        print('stop engine')

class Vehicle:
    def __init__(self, type: str, forSale: bool, engine: Engine):
        self.type = type
        self.forSale = forSale
        self.engine = engine

engine = Engine("V6");
vehicle = Vehicle("Car", True, engine)
# Example of compistion where one object calls the public method of aother object passed in to the constructor.
vehicle.engine.startEngine()
```

### Default/Empty Constructors
Below is an example of the default / empty construtors. This class can be instantiated using `enemy = Enemy()`

```python

class Enemy:
    type_of_enemy: str
    health_points: int=10
    attack_damage: int=1

    # This is a default construtor automatically created if no constructor is found. 
    # the below section has been manually added to represetn an example but in actuality this will be hidden for default.
    def __init__(self):
        pass

    def talk(self):
        print(f'I am a {self.type_of_enemy}. Be prepared to fight')

    def walk_forward(self):
        print(f'{self.type_of_enemy} moves close to you')
    
    def attack(self):
        print(f'{self.type_of_enemy} attacks for {self.attack_damage}')

#instantiating the Class
enemy = Enemy()
enemy.type_of_enemy='Vampires'
enemy.talk()
enemy.walk_forward()
enemy.attack()
```
