## Virtual Environments Overview

- Create a breandnew folder called 'FastAPI'.
- Python already has an included ependency for virtual environments called venv.
- Creating a new FastAPI environmen as a virtual environment.
    - From the root of the project (in this case FastAPI) run the following command.
    ```
    python -m venv fastapienv
    ```
    - Then run `fastapienv\Scripts\activate.bat`
    - after running the script you will notice the virtual environment being activated `(fastapienv) C:\Users\kcabh\Documents\projects\2024\notes\Languages\Python\FastAPI>`
    - To exit out of the venv, type in `deactivate`
- Activating our virtual environment to install our dependencies.
